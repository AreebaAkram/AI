```python
df = pd.read_csv("messy_data.csv")
print(df.head())
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
# 1. remove null values
res = df.dropna() # inplace = True
print(res.isnull().sum())
```

```python
# 2. Replace Null values
dataset = pd.read_csv("messy_data.csv")
# print(dataset.isnull().sum())
res = dataset.fillna({"Age" : 20, "Marks" : 32.0}) #Without it, fillna() returns a new DataFrame, so you would need to assign the result
print(res.isnull().sum())
```

```python
# Replace with mean median mode
dataset = pd.read_csv("messy_data.csv") 
dataset["Age"] = pd.to_numeric(dataset["Age"], errors = "coerce") # to be able to take mean as alphabets converted into NaN
age_mean = dataset["Age"].mean()
print(age_mean)
res = dataset.fillna({"Age" : round(age_mean) })
# print(res.isnull().sum())
# print(res["Age"])

# res["Age"] = res["Age"].round().astype("Int64")
print(res["Age"])
```

```python
# cleaning wrong data
dataset = pd.read_csv("messy_data.csv") 
dataset["Age"] = pd.to_numeric(dataset["Age"], errors = "coerce")
for x in dataset.index:
    if dataset.loc[x, "Age"] < 0 : #retrieve row's age
        # dataset.loc[x, "Age"] = 20
        dataset.drop(x, inplace = True)
print(dataset["Age"])
```

```python
# Remove duplicate rows
dataset = pd.read_csv("duplicatedSet.csv")
# print(dataset.duplicated())
dataset.drop_duplicates(inplace = True)
print(dataset.duplicated())
```

```python
# pandas corelation
dataset = pd.read_csv("messy_data.csv") 
dataset["Age"] = pd.to_numeric(dataset["Age"], errors="coerce")
dataset["Marks"] = pd.to_numeric(dataset["Marks"], errors="coerce")

print(dataset["Age"].corr(dataset["Marks"]))  # how closely they are related to each other
```
