```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([4,6,8,10,11])
plt.plot(x,y, color = 'red' , linestyle = '--', marker = 's')
plt.xlabel ("X axis")
plt.ylabel("y label")
plt.fill_between(x,y, color= 'blue')
plt.title  ("Robotics")
plt.plot()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1, 4,8])
y = np.array([2,7,15])
a= np.array([1,2,3,4,5])
b = np.array([1,4,9,16,25])
plt.plot(a,b, color = "green", marker = 's', label = "input")
plt.plot(x,y, color= "red",  marker = 's', label = "outcome")
plt.xlabel ("X axis")
plt.ylabel("y axis")
plt.fill_between(x,y, color= 'lightpink')
plt.fill_between(a,b, color= 'skyblue')
plt.title  ("Graph of Robot Localization")
plt.legend()
plt.plot()
plt.show()
```
