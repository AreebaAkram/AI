```python
import tensorflow as tf
import numpy as np


# age, bmi, b.p, cjolesterol level
data = np.array(
    [
        [25, 22,120,180],
        [45, 28,135,190],
        [35,32,140,220],
        [50, 30, 150,250],
        [60, 35, 160, 260],
        [55, 29, 145, 240],
        [40, 25, 130, 200]
    ]
)

labels = np.array([0,0,1,1,1,1,0])

model = tf.keras.Sequential([
     tf.keras.layers.Dense(units=1, input_shape=(4,), activation='sigmoid')
])

model.compile(optimizer = 'adam' , loss = "binary_crossentropy", metrics = ["accuracy"])

model.fit(data, labels, epochs=100)

input_data = np.array([[45,33,155,245]])

prediction = model.predict(input_data)
print("predictions: ", prediction[0][0])
if prediction[0][0] > 0.5:
    print("The model predicts that the patient has that disease.")
else:
    print("The model predicts that the patient does not have disease")
```

```python
# Task : text classification

import tensorflow as tf
import numpy as np

iris_data = np.array([
    [5.1, 3.5, 1.4, 0.2],
    [4.9, 3.0, 1.4, 0.2],
    [6.7, 3.1, 4.7, 1.5],
    [5.6, 2.8, 4.9, 2.0],
    [5.0, 3.6, 1.4, 0.2],
    [6.3, 3.3, 6.0, 2.5],
    [5.1, 3.8, 1.5, 0.3],
])

iris_labels = np.array([1,1,0,0,1,0,1])

iris_model = tf.keras.Sequential(
    [
        tf.keras.layers.Dense(units= 1, input_shape = (4,), activation = "sigmoid")
    ]
)

iris_model.compile(optimizer = "adam", loss = "binary_crossentropy", metrics = ['accuracy'])

print("Training iris setosa classifier....")

iris_model.fit(iris_data, iris_labels, epochs = 50, verbose=0)

iris_input = np.array([[5.0, 3.4, 1.5, 0.2]])

iris_prediction = iris_model.predict(iris_input)
print("Iris setosa prediction (probability):", iris_prediction[0][0])
print("Prediction:", "Iris-setosa" 
      if iris_prediction[0][0] > 0.5 
      else "Not Iris-setosa")
print()
```
