Handling missing data

```python

import pandas as pd

data = pd.read_csv('student-mat.csv')
print(data.head()) #will print first 5 rows ->default
print(data.isnull()) # check for null values
```

```python
print(data.isnull().sum())
```

```python
# drop null values
clean_data = data.dropna()
print(clean_data.isnull().sum())
```

```python
data.fillna({'age' : data['age'].mean()}, inplace = True ) # replace null values with mean
```
