```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['G'],
    'D': ['F', ],
    'E': [],
    'F': ['H'],
    'G': ['F']
}

def get_neighbors(state):
    return graph.get(state, [])

def goal_test(state):
    return state == 'G'  # Change this to your desired goal

def bfs(initial_state, goal_test, get_neighbors):
    node_list = deque([initial_state])
    visited = set()
    traversal_order = []

    while node_list:
        current_state = node_list.popleft()

        if current_state not in visited:
            visited.add(current_state)
        traversal_order.append(current_state)  # Track the order of traversal

        if goal_test(current_state):
            print("Traversal Order:", traversal_order)
            return current_state

        for i in get_neighbors(current_state):
            if i not in visited:
                node_list.append(i)

    print("Traversal Order:", traversal_order)
    return None  # If goal not found

# Example usage
result = bfs('A', goal_test, get_neighbors)
print("Goal state reached:", result)
```
