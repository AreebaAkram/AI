Data Transformation & Feature Engineering

1 Feature Scaling : 1. Standardization & Normalization 2. Categorical Data Encoding

```python
import pandas as pd

data = pd.read_csv('student-mat.csv')
print(data.head()) #will print first 5 rows ->default
```

```python
# Normalization -> MinMaxScaling 

from sklearn.preprocessing import MinMaxScaler


scaler = MinMaxScaler()

#scaled_data = scaler.fit_trasform(data) -- for whole data
data['age'] = scaler.fit_transform(data[['age']]) # for specific columns

print(data['age'].head())
print(data['age'].describe())

# After scaling
print("Min:", data['age'].min())
print("Max:", data['age'].max())
```

```python
# Standardization -> StandardScaling
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler()
data['Mother_edu'] = scaler.fit_transform(data[['Mother_edu']])
print(data['Mother_edu'].head())
print(data['Mother_edu'].describe())

```

```python
# Categorical Data Encoding -> Label encoding
from sklearn.preprocessing import LabelEncoder

data = ['High', 'Medium', 'Low', 'High', 'Low']
encoder = LabelEncoder()
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```

```python
# One Hot Encoding

data = ['Civic', 'Toyota', 'Kia', 'Lancruiser', 'Civic', 'Kia']
df = pd.DataFrame(data, columns = ['cars'])
encoded_data = pd.get_dummies(df['cars'])
print(encoded_data)
```

```python
# Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder

data = [['First'], ['Second'], ['First'], ['Third']]

encoder = OrdinalEncoder(categories = [['First', 'Second', 'Third']])
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```
