Descriptive Statistic

```python
import pandas as pd

data = pd.read_csv('student-mat.csv')
print(data.head()) #will print first 5 rows ->default
```

```python
import numpy as np
mean = data['age'].mean()
mode = data['age'].mode()
median = np.median(data['age'])
print("mean" , mean)
print("mode" , mode)
print("median" , median)
```

```python
import statistics
results = statistics.stdev(data['age']) # value of dispersion of data
print(results)
```
