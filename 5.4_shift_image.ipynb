```python
import cv2
import numpy as np

image =cv2.imread("a.png")

# [1,0] -> for x-axis  [0,1] -> for y-axis
# thirs num is for transformation of how much pixels
matrix = np.float32([[1,0,50],[0,1,50]])


shifted_img = cv2.warpAffine(image, matrix, (image.shape[1], image.shape[0]))
cv2.imwrite("shifted_image.png", shifted_img)
cv2.imshow("shifted_image", shifted_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
