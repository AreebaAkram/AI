```python
import cv2
import numpy as np

image =cv2.imread("a.png")

start_point = (560,560)
end_point = (70,70)

color =(0,255,0)
thickness = 1
image_with_line = cv2.line(image, start_point, end_point, color, thickness)

cv2.imshow("Image with Line", image_with_line)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
