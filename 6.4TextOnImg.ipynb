```python
import cv2
import numpy as np
image = cv2.imread("nn.JFIF")
text = "This is dog"
position = (0,100)
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 5
color = (255,255,255)
thickness = 2
image_with_text = cv2.putText(image, text, position, font, font_scale, color, thickness)
cv2.imshow("image with txt", image_with_text)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
