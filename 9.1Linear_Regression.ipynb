```python
import numpy as np
from sklearn.linear_model import LinearRegression
exp = np.array([[1], [3] , [5], [7] , [9] ] )
sal = np.array([70000, 75000, 80000, 85000 , 90000])
model = LinearRegression().fit(exp, sal)

predictions = model.predict([[10]])
print("prediction for 10 years: " , predictions[0])

#coeffecient a1
print(model.coef_)

#intercept a0
print(model.intercept_)
```

```python
#use kaggle data set, modify its values, remove unnnecssary values, and train model on that dataset
import numpy as np
experience = np.array([[1], [3] , [5], [7] , [9] ] )
salary = np.array([70000, 75000, 80000, 85000 , 90000])

mean_x = np.mean(experience)
mean_y = np.mean(salary)

numerator = np.sum((experience.flatten() - mean_x)* (salary - mean_y))
denominator = np.sum((experience.flatten() - mean_x)**2)

m = numerator / denominator
b = mean_y - m *mean_x

print("(slope: " , m)
print("intercept " , b)

new_exp = np.array([[11]])  
pre_sal = m * new_exp + b
print("Predicted Salary for experience 11 years:", pre_sal[0][0])
```

```python
#multiple feature with regression model
import numpy as np
from sklearn.linear_model import LinearRegression

#[exp, job role] 0 -> lab assistant 1 -> associate prof 
exp = np.array([ [1,0], [1,1], [2,0], [2,1] , [4,0], [4,1] ])
sal = np.array([ 50000, 80000 , 70000, 100000, 110000 , 120000])

model = LinearRegression().fit(exp, sal)

predictions = model.predict([[7,1]])
print("prediction for 7 years of Associate prof: " , predictions[0])

#coeffecient a1
print(model.coef_)

#intercept a0
print(model.intercept_)

```

```python
#Loss function

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

exp = np.array([[1], [3] , [5], [7] , [9] ] )
sal = np.array([70000, 75000, 80000, 85000 , 90000]) # change some value to check for loss
model = LinearRegression().fit(exp, sal)

prediction = model.predict([[10]])
print("prediction for 10 years: " , prediction[0])

predictions = model.predict(exp) #this contains all the exp values
print(predictions)

loss = mean_squared_error(sal, predictions)
print("mean squared error: " , loss)

#coeffecient a1
print(model.coef_)

#intercept a0
print(model.intercept_)
```

```python

import numpy as np
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error

exp = np.array([[1], [3] , [5], [7] , [9] ] )
sal = np.array([70000, 75000, 80000, 85000 , 90000]) # change some value to check for loss

model = SGDRegressor(learning_rate = "constant", eta0 = 0.01, max_iter = 1000)
model.fit(exp, sal)

prediction = model.predict([[10]])
print("prediction for 10 years: " , prediction[0])

predictions = model.predict(exp) #this contains all the exp values
print(predictions)

loss = mean_squared_error(sal, predictions)
print("mean squared error: " , loss)

#coeffecient a1
print(model.coef_)

#intercept a0
print(model.intercept_)
```
