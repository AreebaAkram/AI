```python
import pandas as pd
import matplotlib.pyplot as plt

dataset = pd.read_csv("retail_sales_data.csv")

dataset["OrderDate"] = pd.to_datetime(dataset["OrderDate"])

plt.plot(dataset["OrderDate"], dataset["Revenue"], "o", label="Revenue")

plt.xlabel("Order Date")
plt.ylabel("Revenue")
plt.title("Revenue Over Time", fontsize=20, loc="left")

plt.xticks(rotation=45) #rotate the labels
plt.tight_layout() #make adjutsmnets in layout
plt.legend()
plt.show()
```

```python
ypoints = np.array([4, 9, 6, 15])
y1 = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle = 'dotted', color="r", linewidth = "10") #ls for linestyle
plt.plot(y1)
plt.grid(axis = 'x') #all line properties are also eligible for this
plt.show()
```

```python
# displaying two plots

#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 1, 1) #no of rows, no. of col, and the plot loc
plt.plot(x,y)

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 1, 2)
plt.plot(x,y)
plt.suptitle("MY SHOP")
plt.show()
```

```python
# scatter plot
dataset['OrderDate'] = pd.to_datetime(dataset['OrderDate'])

plt.scatter(dataset.Quantity , dataset.Revenue, marker = "o", color ="r")
plt.xlabel('Order Date')
plt.ylabel('Revenue')

plt.scatter(dataset.Revenue , dataset.Profit, marker = "o")
plt.title("Revenue Over Time", fontsize = "20", loc = "left")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Bar chat
ypoints = np.array([10,11,14,5,13, 15])
xpoints = np.array(["Bus", "Van", "Auto", "Bike", "Bicycle", "Cars"])
plt.bar(xpoints, ypoints, color ="m", width = 0.5) #To compare values across different categories 
plt.show()
```

```python
#Histogram ->showing frequency distributions.

ypoints = np.random.normal(100,5,150)
plt.hist(ypoints)
plt.show()
```

```python
# pie chart
y = np.array([35, 25, 25, 15])
#x/sum(x) ->for each part of chart
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]
plt.pie(y, labels = mylabels, startangle = 90, explode =  myexplode)
plt.legend() #To add a list of explanation for each wedge
plt.show() 
```
