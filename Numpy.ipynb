```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
arrSum = arr1 + arr2
arrsums = np.sum(arr1) # sum of all elements in a single scalar
arrMulti = np.dot(arr1, arr2)  #algebraic multiplication -> dot product
arrmultiply = arr1 * arr2  #element wise multiplication
arrmultip = np.prod(arr1) # product of all element in a single scalar
print(f" Addition of two arrays: {arrSum}")
print(f" Sum of all elements in an array: {arrsums}")
print(f" Dot product of two vectors: {arrMulti}")
print(f" Elememnt wise multiplication: {arrmultiply}")
print(f" Product of all elements in an array: {arrmultip}")
```

```python
# BroadCasting

arr1 = np.array([1,2,3,4])
result = arr1 + 10   #it is strected as [10,10,10,10]
print(result)

arr2 = np.array([[1,2,3], [4,5,6]])

# row wise
arr3 = np.array([10,20,30])

# col wise
arr4 = np.array([[10],   
    [11]])
results = arr2 + arr3
res = arr2 + arr4

print(results)
print(res)
```

```python
# Vectorization: Vectorization in NumPy refers to applying operations on entire arrays without using explicit loops.

arr1 = np.array([1,2,3,4])
result = arr1 +10
print(result)

num = 2
print(arr1<num)

res = arr1 * 3
print(res)

vec = np.vectorize(lambda x: x**2 +2*x +1)
results = vec(arr1)
print(f" custom function: {results}")

arr_mean = np.mean(arr1)
print(f" Mean : {arr_mean}")

arr_median = np.median(arr1)
print(f" Median : {arr_median}")

std = np.std(arr1)
print(std)
```

```python
# element wise operations

arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
subtraction = arr1-arr2
print(subtraction)
div = arr1 / arr2
print(div)
```

```python
arr1 = np.array([[1,2],[5,6]])
arr2 = np.array([[1,0],[2,3]])  #no. of rows in first must be equal to num of columns in 2nd

result = np.dot(arr1, arr2)
print(result)
```
