Intro to Numpy

```python
import numpy as np

array_1 = ([1,2,3,4])
array_2 = ([4,3,2,1])

print(array_1 + array_2)
```

```python
array_3 = np.array([1,2,3,4])
array_4 = np.array([4,3,2,1])

print(array_3 + array_4)
```

```python
matrix_1 = ([2,4],
           [6,8])
matrix_2 = ([12,6],
           [7,9])
print(np.dot(matrix_1, matrix_2))
print(np.add(matrix_1, matrix_2))
```
