Python Basics

```python
# Variables
a = 5
b = 6.2
name = "areeba"
print(name)
```

```python
# operators
print (a+b)
x = a**2
print(x)
```

```python
# Control Flow
if a > b:
    print(f"{a} is greater than {b}")
else:
     print(f"{a} is smaller than {b}")

for i in range(5):
    print (i)

for i in (0,...,5):
    print (i)
j = 0
while j < 5:
    print(j)
    j +=1
```

```python
# Lists
numbers = [1,2,3,4,5]
print(numbers[4])
numbers.pop(2)
numbers.append([10])
print(numbers)
```

```python
# Dictionaires
student = {"name": "areeba", "age": 20, "is_student": True}
print(student["name"])
```
