```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
exp= np.array([[1, 0],[3, 0],[5, 1],[7, 0],[9, 1]])
salary = np.array([70000, 75000, 80000, 85000, 90000])
model = LinearRegression().fit(exp, salary)
```

```python
prediction = model.predict([[10, 1]])
print("prediction for 10 years:", prediction[0])
```

```python
print(model.coef_
```

```python
print(model.intercept_)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
exp= np.array([
    [1, 23, 16, 0],
    [3, 22, 18, 0],
    [5, 20, 16, 1],
    [7, 26, 18, 0],
    [9, 25, 16, 1]])
salary = np.array([70000, 75000, 80000, 85000, 90000])
model = LinearRegression().fit(exp, salary)
prediction = model.predict([[11, 20, 16, 1]])
print("prediction for 11 years:", prediction[0])
print(model.coef_)
print(model.intercept_)
```
