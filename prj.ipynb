```python
import torch
import imageio
from diffusers import TextToVideoZeroPipeline

model_id = "Lightricks/LTX-Video"
pipe = TextToVideoZeroPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
).to("cpu")

prompt = "A robot teaching a classroom of students about recycling"
negative_prompt = "Distorted, blurry, low resolution, static, disfigured, unnatural colors"

result = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=50, 
    guidance_scale=7.5,    
    motion_field_strength_x=12,
    motion_field_strength_y=12
).images
result = [(r * 255).astype("uint8") for r in result]
imageio.mimsave("output_video.mp4", result, fps=8)

print("Video generated and saved as 'output_video.mp4'")
```
